# Holstein–Jaynes–Cummings spectrum

Cleaned and organized for reproducible execution from the repository root.


```python
import numpy as np
from qutip import tensor, qeye, sigmaz, sigmap, sigmam, destroy
```

```python
# Parameters
default_params = {
    'omega_q': 1.0,      # qubit splitting
    'omega_r': 1.0,      # resonator frequency
    'g_z': 0.1,          # coupling strength for sigma_z(a + a.dag())
    'g_x': 0.1,          # coupling strength for JC term sigma_+ a + sigma_- a.dag()
    'N':    20           # number of resonator Fock states
}
```

```python
# 1. Unpack parameters
omega_q = default_params['omega_q']
omega_r = default_params['omega_r']
g_z     = default_params['g_z']
g_x     = default_params['g_x']
N       = default_params['N']
```

```python
# 2. Define subsystem operators
# Qubit: 2-level system
i_q = qeye(2)
s_z = sigmaz()
s_p = sigmap()
s_m = sigmam()
# Resonator: bosonic mode

i_r = qeye(N)
a   = destroy(N)
a_d = a.dag()
```

```python
# 3. Tensor to full Hilbert space
I_qI_r = tensor(i_q, i_r)
Sz_I   = tensor(s_z, i_r)
Sp_a   = tensor(s_p, a)
Sm_ad  = tensor(s_m, a_d)
aI     = tensor(i_q, a)
adI    = tensor(i_q, a_d)
```

```python
# 4. Build Hamiltonians
# Qubit Hamiltonian: (omega_q/2) * sigma_z
H_q = (omega_q/2) * Sz_I
# Resonator Hamiltonian: omega_r * a.dag() * a
H_r = omega_r * tensor(i_q, a_d * a)
# Holstein-like coupling: g_z * sigma_z (a + a.dag())
H_z = g_z * tensor(s_z, a + a_d)
# Jaynes–Cummings coupling: g_x * (sigma_+ a + sigma_- a.dag())
H_x = g_x * (Sp_a + Sm_ad)
```

```python
# 5. Total Hamiltonians
total_H_holstein = H_q + H_r + H_z
total_H_jc        = H_q + H_r + H_x
```

```python
# 6. Example: diagonalize Holstein Hamiltonian
eigs_holstein = total_H_holstein.eigenenergies()
```

```python
# 7. Print results
if __name__ == '__main__':
    print('First 10 eigenenergies of Holstein-like model:')
    for n, e in enumerate(eigs_holstein[:10]):
        print(f"  Level {n}: {e:.4f}")
```
